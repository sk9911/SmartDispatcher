```python
import pandas as pd
from regret import stats
from regret.figures.lead_regret import build_lead_regret, FIG_ID
df = pd.read_parquet('tests/fixture_results/regret.parquet')
fig = build_lead_regret(df)
fig.show()
agg = stats.aggregate(df, by=('source', 'lead_h', 'direction'))
print('sources', list(dict.fromkeys(agg['source'])))
print('n_traces', len(fig.data))
print('traces', [(t.name, t.legendgroup, t.xaxis, t.showlegend) for t in fig.data])
fig.write_html('qa/T23/scratch/fig.html', include_plotlyjs=True, div_id=FIG_ID)
print('wrote fig.html')
```

```python
import pandas as pd
from regret.figures.wind_error import build_wind_error, FIG_ID
df = pd.read_parquet('data/results/regret.parquet')
fig = build_wind_error(df)
fig.show()
fc = df[~df['is_truth_optimised']]
print('rows', len(df), 'forecast_plan_rows', len(fc))
print('rows_per_source', fc.groupby('source', sort=False).size().to_dict())
print('traces', [(t.name, len(t.x)) for t in fig.data])
fig.write_html('qa/B6/scratch/wind.html', include_plotlyjs=True, div_id=FIG_ID)
print('wrote wind.html')
```

```python
from regret.figures.lead_regret import build_lead_regret, FIG_ID as LID
from regret.figures.sanity import build_sanity, FIG_ID as SID
lfig = build_lead_regret(df)
lfig.show()
lfig.write_html('qa/B6/scratch/lead.html', include_plotlyjs=True, div_id=LID)
sfig = build_sanity(df)
sfig.write_html('qa/B6/scratch/sanity.html', include_plotlyjs=True, div_id=SID)
print('wrote lead.html sanity.html')
```

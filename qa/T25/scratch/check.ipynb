```python
import pandas as pd
from regret.figures.wind_error import build_wind_error, FIG_ID
df = pd.read_parquet('tests/fixture_results/regret.parquet')
fig = build_wind_error(df)
fig.show()
fc = df[~df['is_truth_optimised']]
print('forecast_plan_rows', len(fc))
print('rows_per_source', fc.groupby('source', sort=False).size().to_dict())
print('traces', [(t.name, len(t.x)) for t in fig.data])
print('wind_error range', fc['wind_error_ms'].min(), fc['wind_error_ms'].max())
print('regret range', fc['regret_fuel_kg'].min(), fc['regret_fuel_kg'].max())
print(fc.groupby('source', sort=False)[['wind_error_ms','regret_fuel_kg']].describe().T.to_string())
print('nan wind_error', int(fc['wind_error_ms'].isna().sum()), 'nan regret', int(fc['regret_fuel_kg'].isna().sum()))
fig.write_html('qa/T25/scratch/fig.html', include_plotlyjs=True, div_id=FIG_ID)
print('wrote fig.html')
```
